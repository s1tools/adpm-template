# Impact of a negligible variable

The example quantifies the output contribution of a variable omitted by the approximate algorithm. The vertical band identifies the algorithm's application domain.

```python
import platform

import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print(f"Python:     {platform.python_version()}")
print(f"NumPy:      {np.__version__}")
print(f"Matplotlib: {matplotlib.__version__}")
```

```python
z = np.linspace(0, 5, 200)
impact = 0.08 * z**2 + 0.05 * z
domain, limit = (1.5, 3.0), 1.0

inside = (z >= domain[0]) & (z <= domain[1])
assert impact[inside].max() < limit

figure, axis = plt.subplots(figsize=(6, 3.5), constrained_layout=True)
axis.plot(z, impact, label="Impact of neglected term")
axis.axvspan(*domain, color="tab:green", alpha=0.2, label="Application domain")
axis.axhline(
    limit, color="tab:red", linestyle="--", label="Acceptance threshold"
)
axis.set(
    xlabel=r"Neglected variable $z$",
    ylabel="Relative impact on output [%]",
    title="Impact of the approximation",
)
axis.grid(alpha=0.25)
axis.legend()
# figure.savefig("../_static/negligible_variable_impact.png", dpi=160)
plt.show()
```

*Example of the impact of a neglected variable within the algorithm application domain.*
